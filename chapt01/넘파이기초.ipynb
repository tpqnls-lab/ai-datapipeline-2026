### Python 리스트(배열) 과 NumPy 배열 비교

- 파이썬에 기존 리스트(배열) 자료구조 존재. 속도적면에서 NumPy가 우세

#### 파이썬으로 배열작업

````python
prices = [10000, 15000, 12000, 30000, 25000] # 5개 가격이 든 배열
discount_prices = []

for price in prices:
    discount_prices.append(price *0.9) # 10% 할인

discount_prices  # 노트북 코드쉘에서 하나만 출력하려면 print() 생략 가능
````

#### NumPy로 배열작업

````python
np_prices = np.array(prices)

np_discount_prices = np_prices * 0.9

np_discount_prices
````

#### 결측값 핸들링

- 실제 데이터에서 값이 비어있는 경우, 이를 결측치/결측값이라고 함
- 일반적으로 `NULL`, 파이썬에서는 `None`, NumPy에서는 `np.nan`로 표현

````python
temperatures = np.array([22.5, 23.0, np.nan, 25.1, 21.8])
````

````python
temperatures
````

````python
# 평균값
temperatures.mean()
````

````python
temperatures.min()
````

````python
np.nanmean(temperatures)
````

````python
np.isnan(temperatures)
````

````python
temperatures[~np.isnan(temperatures)]
````

- 결측치를 정리방법 : 결측치는 제거하거나(건수가 줄어듬), 나머지 평균값을 임의로 기입하거나, 기본통계에서 도출할 수 있는 다른 값들로 채우기

#### 이상치 핸들링

- 일반적인 범위를 벗어나는 비정상 데이터.
- ex) 부산 1월 일별 평균기온에 -90도. 습도99%

````python
temperatures = np.array([22.5, 23.0, np.nan, 25.1, 999.0, 21.8, -99])
````

````python
valid_temps = temperatures[
    (~np.isnan(temperatures)) & # nan은 무조건 제거
    (temperatures >= 15.0) &  # 최저기온이 15도 아래로 내려갈 수 없으므로 제거
    (temperatures <= 50) # 최고기온이 50도 이상 오를 수 없으므로 제거
]

valid_temps
````

- 전처리(preProcessing) - 분석 이전에 이상치,결측치를 없애는 작업

### 실습

#### 기온데이터 전처리

````python
temperatures = np.array([22.5, 23.0, np.nan, 25.1, 999.0, 21.8, -99.0, 24.0, 23.5, np.nan])
temperatures
````

- 결측값 개수 출력
- 결측값, 이상치 제거
- 정상 데이터 출력
- 평균 기온 출력
- 최고 기온 출력
- 최저 기온 출력

````python
# 결측값 출력
np.isnan(temperatures)
````

````python
# 결측값 갯수
np.isnan(temperatures).sum()
````

````python
missing_count = np.isnan(temperatures).sum()
missing_count
````

````python
# 결측값 제거, 이상치 제거, 정상 데이터 출력
valid_temps = temperatures[
    (~np.isnan(temperatures)) &
    (temperatures >= 15.0) &
    (temperatures <= 50.0)
]

valid_temps
````

````python
# 평균 기온 출력
valid_temps.mean()
````

````python
# 최저 기온 출력

valid_temps.min()
````

````python
# 최고 기온 출력

valid_temps.max()
````

#### 매출 데이터 전처리

- 결측값 개수 출력
- 결측값 제외
- 0원 10만원 내의 정상 매출만 남김
- 정상 매출의 평균, 최대값, 최소값 출력
- 정상매출에서 평균이상인 값만 출력

````python
# 파이썬 리스트에서 None 대신 np.nan으로 결측치를 입력하면 NumPy 패키지 연산이 가능
raw_sales = [12000, 15000, np.nan, 30000, -5000, 25000, 999999, 18000, np.nan, 22000]

raw_sales
````

````python
# NumPy 배열로 변경
np_raw_sales = np.array(raw_sales)

np_raw_sales
````

````python
raw_sales
````

````python
# 결측값 확인
np.isnan(np_raw_sales)
````

````python
np.isnan(np_raw_sales).sum()
````

````python
np_raw_sales[
    (~np.isnan(np_raw_sales)) &
    (np_raw_sales >= 0) &
    (np_raw_sales <= 100000)
]
````

````python
valid_sales = np_raw_sales[
    (~np.isnan(np_raw_sales)) &
    (np_raw_sales >= 0) &
    (np_raw_sales <= 100000)
]

valid_sales
````

````python
# 평균매출액, 평균매출액 mean_sales담아놓기
mean_sales = valid_sales.mean()
mean_sales
````

````python
valid_sales.min()
````

````python
valid_sales.max()
````

````python
# 평균 매출액보다 높음 매출 찾기
valid_sales[valid_sales > mean_sales]
````
