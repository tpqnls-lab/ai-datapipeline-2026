##### 막대그래프

- 항목별(카테고리) 크기를 비교할 때 사용

```python
df.groupby('category')['total_price'].sum().reset_index()
```

```python
cate_sales = df.groupby('category')['total_price'].sum().reset_index()
```

```python
# x축
cate_sales['category']
```

```python
# y축
cate_sales['total_price']
```

```python
plt.figure(figsize=(10, 5))
# 막대그래프 plt.bar()
plt.bar(x=cate_sales['category'], height=cate_sales['total_price'])
plt.title('항목별 판매금액', fontsize=15, fontweight='bold')
plt.xlabel('카테고리')
plt.ylabel('판매금액')

plt.ticklabel_format(style='plain', axis='y') # y축을 판매금액 그대로 출력
plt.show()
```

##### 선 그래프

- 날짜별 매출 계산

```python
df.groupby('order_date')['total_price'].sum().reset_index()
```

```python
daily_sales = df.groupby('order_date')['total_price'].sum().reset_index()
```

```python
plt.figure(figsize=(10, 5))
plt.plot(daily_sales['order_date'], daily_sales['total_price'], 
         marker='o', 
         markersize=7, 
         linewidth=2.5,
         color= "#92C239")
plt.title('일자별 판매금액')
plt.ylabel('판매금액')
plt.ticklabel_format(style='plain', axis='y')

# 실제 데이터가 있는 위치의 x좌표 레이블 표식
plt.xticks(
    daily_sales['order_date'],
    daily_sales['order_date']. to_string('%m-%d')
)

plt.show()
```

##### 히스토그램

- 표준편차, 분포와 관계있는 그래프
- 숫자 데이터의 분포를 확인할때 사용

```python
plt.figure(figsize=(10, 5))
# 히스토그램
plt.hist(df['total_price'], bins=5) # bins 구간개수
plt.title('구매금액 분포도', fontsize=15)
plt.xlabel('구매금액 분포')
plt.ylabel('갯수')

plt.ticklabel_format(style='plain', axis='x')
plt.show()
```

##### 산점도

- 두 숫자 데이터간의 관계를 확인 그래프
- 데이터분석시 가장 많이 사용

```python
plt.figure(figsize=(10, 5))

# 산점도
plt.scatter(x=df['quantity'], y=df['total_price'])
plt.title('구매수량,금액간의 관계도', fontsize=15)
plt.xlabel('구매갯수')
plt.ylabel('구매금액')

plt.ticklabel_format(style='plain', axis='y')
plt.show()
```

##### 박스플롯

- 데이터의 이상치와 분포를 같이 확인할때 사용

```python
plt.figure(figsize=(10, 5))

# 박스플롯
plt.boxplot(df['total_price'])
plt.title('구매금액 분포와 이상치', fontsize=15)
plt.ylabel('구매금액')

plt.ticklabel_format(style='plain', axis='y')
plt.show()
```

- 선 그래프 마커위치에 데이터 표시

```python
plt.figure(figsize=(10, 5))
plt.plot(daily_sales['order_date'], daily_sales['total_price'], 
         marker='o', 
         markersize=10, 
         linewidth=2.5, 
         color="#EF5A1045")

# 마커위치에 실제 데이터 표시
for x, y in zip(daily_sales['order_date'], daily_sales['total_price']):
    plt.text(
        x, y,
        f"{y:,}",  # y값 = 금액
        ha='left',  # 가로축 정렬 left, center, right  - 정렬 이름 반대
        va='bottom'   # 세로축 정렬 top, center, bottom - 정렬 이름 반대
    )

plt.title('일자별 판매금액')
plt.ylabel('판매금액')
plt.ticklabel_format(style='plain', axis='y')

# 실제 데이터가 있는 위치의 x좌표 레이블 표시
# plt.xticks(
#     daily_sales['order_date'],
#     daily_sales['order_date'].to_string('%y%m%d')  # 텍스트 파일 생성
# )
plt.xticks(
    daily_sales['order_date'],
    daily_sales['order_date'].dt.strftime('%y-%m-%d')
)
plt.show()
```

##### Seaborn으로 손쉽게 차트 구현
- Matplotlib은 판다스 DF를 바로 사용못함
- Seaborn은 DF를 매개변수로 활용가능
- Matplotlib은 x, y(height) 값을 시리즈로 만들어서 전달
- Seaborn은 데이터프레임 그대로 data 속성에 전달

```python
plt.figure(figsize=(10, 5))
# 막대그래프 plt.bar() -> Seaborn barplot으로 변경
sns.barplot(data=cate_sales, x='category', y='total_price')
plt.title('항목별 판매금액', fontsize=15, fontweight='bold')
plt.xlabel('카테고리')
plt.ylabel('판매금액')

plt.ticklabel_format(style='plain', axis='y') # y축을 판매금액 그대로 출력. 2.5 1e6 과 같이 출력
plt.show()
```

##### Seaborn Heatmap
- 숫자 데이터 분포를 색상으로 표현하는 그래프

```python
# 구매수량, 단가, 구매금액 간의 상관관계 
df[['quantity', 'unit_price', 'total_price']].corr()
```

```python
corr = df[['quantity', 'unit_price', 'total_price']].corr()
```

```python
plt.figure(figsize=(10, 5))
# 히트맵
sns.heatmap(data=corr, annot=True, cmap='Blues')
plt.title('구매수량과 단가 간의 상관관계')
plt.show()
```
