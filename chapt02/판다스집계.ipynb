## 판다스 집계

### 집계

- 계산, 통계, 그룹핑으로 원데이터에서 필요한 수치적으로 필요한 값을 재가공

```python
import numpy as np
import pandas as pd
```

```python
orders = pd.DataFrame({
    "order_id": [1, 2, 3, 4, 5, 6],
    "customer": ["홍길동", "김철수", "이영희", "홍길동", "박민수", "김철수"],
    "product_id": ["P001", "P002", "P001", "P003", "P002", "P003"],
    "quantity": [2, 1, 3, 1, 2, 4],
    "unit_price": [30000, 15000, 30000, 8000, 15000, 8000],
    "order_date": [
        "2026-09-01",
        "2026-09-01",
        "2026-09-02",
        "2026-09-02",
        "2026-09-03",
        "2026-09-03"
    ]
})
```

```python
# 주문번호, 고객명, 상품ID, 수량, 단가, 주문일자
orders
```

#### 파생 컬럼 생성

- 단가*수량 -> 주문금액 컬럼

```python
orders['total_price'] = orders['quantity'] * orders['unit_price']
```

```python
orders
```

- 필터링 후 파생컬럼 생성

```python
orders.loc[condition]
```

#### apply() 함수 사용

- apply() 함수 내에 사용할 사용자 정의 함수
- 학점 등 여러 구분으로 나눠지는 도메인 값의 컬럼을 생성할 때 유용

```python
def get_order_grade(total_price):
    if total_price >= 80000:
        return 'A'
    elif total_price >= 30000:
        return 'B'
    else:
        return 'C'
```

```python
get_order_grade(20000)
```

```python
# total_price 컬럼의 값을 사용해서 get_order_grade() 함수를 각 행별로 호출해서 grade 컬럼을 생성
orders['grade'] = orders['total_price'].apply(get_order_grade)
```

#### CONCAT

- 데이터 합치기

```python
week1 = pd.DataFrame({
    "order_id": [1, 2],
    "customer": ["홍길동", "김철수"],
    "amount": [60000, 15000]
})

week2 = pd.DataFrame({
    "order_id": [3, 4],
    "customer": ["이영희", "박민수"],
    "amount": [90000, 30000]
})
```

```python
week1
```

```python
# 이전 인덱스가 그대로 반영
pd.concat([week1, week2])
```

```python
# 일반적으로 가장 많이 사용하는 방법
pd.concat([week1, week2], ignore_index=True)
```

#### Merge

- DB Join과 동일

```python
products = pd.DataFrame({
    "product_id": ["P001", "P002", "P003"],
    "product_name": ["키보드", "마우스", "USB"],
    "category": ["주변기기", "주변기기", "저장장치"]
})
```

```python
orders
```

```python
pd.merge(orders, products, on='product_id')
```

- how : inner, left, right 가 99% 사용

```python
pd.merge(orders, products, on='product_id', how='left')
```

```python
products.loc[3] = {'product_id': 'P004', 'product_name': '모니터', 'category': '디스플레이'}
```

```python
products
```

```python
pd.merge(orders, products, on='product_id', how='right')
```

#### group by 집계
- DB group by 와 동일

```python
orders.groupby('customer')
```

```python
orders.groupby('customer')['total_price'].sum()
```

```python
# 데이터프레임으로 변환
orders.groupby('customer')['total_price'].sum().reset_index()
```

#### 여러 통계 한 번에 계산

```python
# 고객별로 기본통계
orders.groupby('customer')['total_price'].agg(
    ['count', 'sum', 'mean', 'max', 'min']
)
```

```python
orders.groupby('customer')['total_price'].sum().reset_index().describe()
```

```python
# 통계치 CSV 파일 저장
orders.groupby('customer')['total_price'].agg(
    ['count', 'sum', 'mean', 'max', 'min']
).to_csv('../고객별구매통계.csv', encoding='utf-8')
```
